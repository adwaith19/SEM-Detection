Collect all the .tif files and add them into a new folder

```python
import os
import shutil

source_dir = '/content/extracted_SEM_analysis'
dest_dir = '/content/collected_tif_images'

# Create the destination directory if it doesn't exist
os.makedirs(dest_dir, exist_ok=True)

found_files_count = 0

# Keep track of existing filenames to handle duplicates
existing_files = set()

# Walk through the source directory
for root, _, files in os.walk(source_dir):
    for file in files:
        if file.lower().endswith('.tif'):
            source_file_path = os.path.join(root, file)

            base_name, extension = os.path.splitext(file)
            destination_file_name = file
            counter = 1

            # Generate unique filename if one with the same name already exists
            while destination_file_name in existing_files:
                destination_file_name = f"{base_name}_{counter}{extension}"
                counter += 1

            destination_file_path = os.path.join(dest_dir, destination_file_name)

            try:
                shutil.copy2(source_file_path, destination_file_path)
                existing_files.add(destination_file_name) # Add the new unique name to the set
                found_files_count += 1
                print(f"Copied: {source_file_path} to {destination_file_path}")
            except Exception as e:
                print(f"Error copying '{source_file_path}': {e}")

print(f"\nFinished copying {found_files_count} .tif files to '{dest_dir}'.")
print("Total unique .tif files collected: ", len(existing_files))
```

```python
from PIL import Image
import os

dest_dir = '/content/collected_tif_images'

# Get all .tif files in the destination directory
tif_files = [f for f in os.listdir(dest_dir) if f.lower().endswith('.tif')]

print(f"Found {len(tif_files)} .tif files for conversion.")

# Convert and rename files
for i, tif_file in enumerate(tif_files):
    tif_path = os.path.join(dest_dir, tif_file)
    png_filename = f"{i}.png"
    png_path = os.path.join(dest_dir, png_filename)

    try:
        with Image.open(tif_path) as img:
            img.save(png_path)
        print(f"Converted '{tif_file}' to '{png_filename}'")
    except Exception as e:
        print(f"Error converting '{tif_file}': {e}")

print(f"\nFinished converting and renaming {len(tif_files)} files.")
```

```python
from PIL import Image
import os
from collections import Counter

dest_dir = '/content/collected_tif_images'

# Get all .png files in the destination directory
png_files = [f for f in os.listdir(dest_dir) if f.lower().endswith('.png')]

if not png_files:
    print(f"No .png files found in '{dest_dir}'.")
else:
    print(f"Found {len(png_files)} .png files to analyze.")

    resolutions = []
    for png_file in png_files:
        png_path = os.path.join(dest_dir, png_file)
        try:
            with Image.open(png_path) as img:
                resolutions.append(img.size) # img.size returns (width, height)
        except Exception as e:
            print(f"Error opening image '{png_file}': {e}")

    # Count occurrences of each resolution
    resolution_counts = Counter(resolutions)

    print("\nImage Resolution Categories and Counts:")
    if resolution_counts:
        for res, count in resolution_counts.most_common():
            print(f"  {res[0]}x{res[1]}: {count} images")
    else:
        print("No valid image resolutions found.")
```

```python
from PIL import Image
import os

dest_dir = '/content/collected_tif_images'

# Get all .png files in the destination directory
png_files = [f for f in os.listdir(dest_dir) if f.lower().endswith('.png')]

print(f"Found {len(png_files)} .png files to crop.")

crop_width = 768
crop_height = 768

cropped_count = 0
for png_file in png_files:
    png_path = os.path.join(dest_dir, png_file)
    try:
        with Image.open(png_path) as img:
            # Define the crop box (left, upper, right, lower)
            # To crop 756x756 from the top-left corner:
            # left = 0, upper = 0, right = 756, lower = 756
            crop_box = (0, 0, crop_width, crop_height)

            # Perform the crop
            cropped_img = img.crop(crop_box)

            # Save the cropped image, overwriting the original
            cropped_img.save(png_path)
            cropped_count += 1
            # print(f"Cropped and saved '{png_file}' to {crop_width}x{crop_height}")
    except Exception as e:
        print(f"Error cropping or saving '{png_file}': {e}")

print(f"\nFinished cropping {cropped_count} images to {crop_width}x{crop_height}.")
```

```python
from PIL import Image
import os

dest_dir = '/content/collected_tif_images'
original_images_dir = os.path.join(dest_dir, 'original_768x768')

png_files = [f for f in os.listdir(dest_dir) if f.lower().endswith('.png') and f != '.ipynb_checkpoints']

print(f"Found {len(png_files)} .png files (768x768) to process.")

crop_size = 256
num_tiles_per_side = 3 # 768 / 256 = 3

cropped_images_generated = 0

for png_file in png_files:
    png_path = os.path.join(dest_dir, png_file)
    try:
        with Image.open(png_path) as img:
            # Ensure image is 768x768 before cropping
            if img.size != (768, 768):
                print(f"Warning: '{png_file}' is not 768x768. Skipping cropping for this image.")
                continue

            # Generate 9 tiles (3x3 grid)
            for i in range(num_tiles_per_side):
                for j in range(num_tiles_per_side):
                    left = j * crop_size
                    upper = i * crop_size
                    right = left + crop_size
                    lower = upper + crop_size

                    crop_box = (left, upper, right, lower)
                    cropped_img = img.crop(crop_box)

                    # Save the cropped image with a temporary name for now
                    # We'll rename them all in a later step
                    tile_name = f"temp_{os.path.splitext(png_file)[0]}_tile_{i}_{j}.png"
                    cropped_img.save(os.path.join(dest_dir, tile_name))
                    cropped_images_generated += 1

    except Exception as e:
        print(f"Error processing '{png_file}': {e}")

print(f"\nFinished generating {cropped_images_generated} new 256x256 images from original 768x768 images.")
```

```python
import os
import shutil
from PIL import Image

dest_dir = '/content/collected_tif_images'
original_768_dir = os.path.join(dest_dir, 'original_768x768')
resized_256_from_original_dir = os.path.join(dest_dir, 'resized_256x256_from_originals')

# Create the new directory for resized copies
os.makedirs(resized_256_from_original_dir, exist_ok=True)
print(f"Created directory: '{resized_256_from_original_dir}'")

# Copy images from original_768x768 to the new directory
original_png_files = [f for f in os.listdir(original_768_dir) if f.lower().endswith('.png')]

print(f"Copying {len(original_png_files)} images from '{original_768_dir}' to '{resized_256_from_original_dir}'.")

for png_file in original_png_files:
    source_path = os.path.join(original_768_dir, png_file)
    destination_path = os.path.join(resized_256_from_original_dir, png_file)
    try:
        shutil.copy2(source_path, destination_path)
    except Exception as e:
        print(f"Error copying '{png_file}': {e}")

print("Finished copying images.")

# Now, resize the copied images in the new directory
png_files_to_resize = [f for f in os.listdir(resized_256_from_original_dir) if f.lower().endswith('.png')]

target_size = (256, 256)
resized_count = 0

print(f"Resizing {len(png_files_to_resize)} images in '{resized_256_from_original_dir}' to {target_size[0]}x{target_size[1]}.")

for png_file in png_files_to_resize:
    png_path = os.path.join(resized_256_from_original_dir, png_file)
    try:
        with Image.open(png_path) as img:
            resized_img = img.resize(target_size, Image.LANCZOS)
            resized_img.save(png_path)
            resized_count += 1
    except Exception as e:
        print(f"Error resizing or saving '{png_file}': {e}")

print(f"\nFinished resizing {resized_count} images to {target_size[0]}x{target_size[1]} in '{resized_256_from_original_dir}'.")
```
